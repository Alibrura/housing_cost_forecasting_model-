Рассмотрев  признак `homeFacts` мы смогли создать новые признаки такие как:
- `year_built` - год постройки недвижитмости
- `remodeled_year` - год рекнострукции недвижимости 
- `heating` - тип отопления 
- `cooling` - тип охлаждения
- `parking` - тип парковки
- `lotsize` -  это размер земельного участка, на котором расположена недвижимость.
- `price/sqft` - это стоимость одного квадратного фута площади объекта. 

````python
def parse_homefacts(text):
    # создаем словарь из строки
    dt = ast.literal_eval(text)

    # Извлекаем список фактов (ключ atAGlanceFacts может отсутствовать)
    facts_list = dt.get('atAGlanceFacts', [])
    
    result = {}
    for item in facts_list:
        label = item.get('factLabel')
        value = item.get('factValue')
        if label is not None:
            # Нормализуем название (убираем пробелы, приводим к нижнему регистру)
            norm_label = label.lower().replace(' ', '_')
            result[norm_label] = value
    return result
        

# Применяем функцию, получаем Series словарей
facts_series = house_df['homeFacts'].apply(parse_homefacts)

# Превращаем в DataFrame
facts_df = pd.json_normalize(facts_series)
# display(facts_df)

# Склеиваем с исходным DataFrame
house_df = pd.concat([house_df, facts_df], axis=1)

# Удаляем исходный признак
house_df.drop('homeFacts', axis=1, inplace=True)
display(house_df.iloc[1:5, 13:])
````

````python
def cleaning_rating(ratings:list = []):
    """Функция для обработки рейтинга"""
    # преобразуем рейтинг в числовой формат
    ratings_cleaned = [] 
    for r in ratings:
        # учебное заведение не оценено 
        if (r == 'NR') or (r == 'NA') or (pd.isna(r)) or (r == ''): 
            continue
        if isinstance(r, str) and '/' in r:
            r_clean = r.replace('/10', '').strip()
            # исключение если былло None/10
            if r_clean == 'None' or r_clean == '':
                continue
        else:
            r_clean = r
            
        try:
            r_clean = float(r_clean)
            ratings_cleaned.append(r_clean)
        except ValueError as ex:
            print(ex)
            continue 

    return ratings_cleaned


def cleaning_distance(distances:list = []):
    """Функция для обработки дистанции"""
    distances_cleaned = []
    for d in distances:
        if d == '' or pd.isna(d) or d == 'None':
            continue
        
        if isinstance(d, str) and 'mi' in d:
            d_cleaned = d.replace('mi', '')
    
        else:
            d_cleaned = d
            
        try:
            d_cleaned = float(d_cleaned)
            distances_cleaned.append(d_cleaned)
        except ValueError as ex:
            print(ex)
            continue
        
    return distances_cleaned 


def cleaning_grades(grades:list = []):
    "Функция для обработки уровня образования"
    cleaned_grades = set()

    def grade_split(grade):
        "Функция для разделения grade на две части"
        if 'preschoolto' in grade:
            grade = grade.replace('preschoolto', '') # sting
        elif 'to' in grade:
            grade = grade.split('to') # list
        elif '-' in grade:
            grade = grade.split('-') # list
        elif '–' in grade:
            grade = grade.split('–') # list
        elif grade in ('k', 'pk'):
            grade = [0]
        elif grade in ('n/a', 'na'):
            grade = list()
        else: # если число в sting формате 
            grade = list(grade)
        return grade 
    
        
    for g in grades:
        if isinstance(g, str) and g!='' and g!='None':
            # удаляем пробелы
            g_clean = g.replace(' ', '').lower()
            
            # разделяем по запятой
            if ',' in g_clean:
                g_clean = g_clean.split(',')
                # по отдельности обработаем обе части потом соберем обратнов в list
                left_g = grade_split(g_clean[0])
                right_g = grade_split(g_clean[1])
                g_clean = left_g + right_g
            else: 
                g_clean = grade_split(g_clean)
            
            # если список то приводим все значенния в числовой формат
            if isinstance(g_clean, list):
                for element in g_clean:
                    if element not in ('k', 'pk', 'n/a', 'na'):
                        cleaned_grades.add(int(element)) # переводим в числовой формат

            else: # string
                if g_clean not in ('k', 'pk', 'n/a', 'na'):
                    cleaned_grades.add(int(g_clean))

                    
            
    return cleaned_grades


def classifier_grades(grades_cleaned):
    """Функция для классификации уровня школы на начальная, средняя, страшая"""
    res = {
        'is_junior': False,
        'is_middle': False,
        'is_high': False
    }
    
    for grade in grades_cleaned:
        if grade < 6:
            res['is_junior'] = True
        elif grade < 9:
            res['is_middle'] = True
        elif grade <= 12:
            res['is_high'] = True
            
    return res
       
````

````python
def parse_schools(text:str):
    lst = ast.literal_eval(text)
    # первый элемент списка - словарь с информацией
    dt = lst[0]
    # итоговый словарь 
    result = {}
    # количество школ
    result['schools_total'] = len(dt.get('name', []))
    
    # Извлекаем рейтинги и расстояния
    ratings = dt.get('rating', [])
    distances = dt.get('data', {}).get('Distance', [])
    grades = dt.get('data', {}).get('Grades', [])
    
    # получаем очищенный рейтинг
    ratings_cleaned = cleaning_rating(ratings)      
    # если данных нет
    if len(ratings_cleaned) == 0 :
        # средний рейтинг школ
        result['average_rating_schools'] = np.nan
        # максимальный рейтинг школ
        result['max_rating_schools'] = np.nan
    # если данные есть
    else:   
        # средний рейтинг школ
        result['average_rating_schools'] = np.mean(ratings_cleaned).round(2)
        # максимальный рейтинг школ
        result['max_rating_schools'] = max(ratings_cleaned)
    
    # --------------------------------------------------------------
    
    distances_cleaned = cleaning_distance(distances)
    # если данных нет
    if len(distances_cleaned) == 0:
        # минимальное расстояние до школы
        result['min_distance_school'] = np.nan
        # среднее растояние до школы
        result['average_distance_school'] = np.nan
    # если данные есть    
    else:
        # минимальное расстояние до школы
        result['min_distance_school'] = round(min(distances_cleaned), 2)
        # среднее растояние до школы
        result['average_distance_school'] = np.mean(distances_cleaned).round(2)
    
    # ----------------------------------------------------------------

    grades_cleaned = cleaning_grades(grades)
    # если данных нет
    if len(grades_cleaned) == 0:
        result['has_junior_school'] = np.nan
        result['has_middle_school'] = np.nan
        result['has_high_school'] = np.nan
    # если данные есть    
    else:
        dt = classifier_grades(grades_cleaned)
        result['has_junior_school'] = dt['is_junior']
        result['has_middle_school'] = dt['is_middle']
        result['has_high_school'] = dt['is_high']
        
    return result
    
# display(parse_schools(house_df['schools'].iloc[884]))

schools_series = house_df['schools'].apply(parse_schools)
# преобразуем из series в dataframe
schools_df = pd.json_normalize(schools_series)

# объединяем в единый датасет
house_df = pd.concat([house_df, schools_df], axis=1)
# удаляем исходный признак
house_df.drop('schools', axis=1, inplace=True)

# закодируем бинарные признаки в category
# house_df[[
#     'has_junior_school',
#     'has_middle_school',
#     'has_high_school'
# ]] = house_df[['has_junior_school','has_middle_school','has_high_school']].astype('category')
display(house_df.iloc[1:5, 21:])
````

Вспомогательные функции для взаимодействи с текстовыми признаками:
1. `split_type` - рекурсивная функция предназначенная дла разделения строковых наблюдений по определенным разделителям (separators). По предназначению напоминает стандратный `split`, но она отличается своим рекурсивным подходом, который в свою очередь помогает разделить всю дополнительную информацию содержащуюся в наблюдениях, а после объединить в множество содержащее только уникальные значения. К примеру `type , type , type/type/type`

2. `regex_search` - функция предназначенная для regex поиска по словарю стандартизованных значений и замены на values.

````python
def split_type(type_str: str, separators: list = [',', '/'])->set: # return set
        """Рекурсивная функция для разделения тегов по нескольким разделителям"""
    
        # Базовый случай: если ни одного разделителя нет, возвращаем строку
        if not any(sep in type_str for sep in separators):
            return {type_str.strip()}  # возвращаем множество с одним элементом
        
        # Ищем первый попавшийся разделитель
        result = set()
        for sep in separators:
            if sep in type_str:
                # Разделяем по первому найденному разделителю
                parts = type_str.split(sep)
                # Для каждой части рекурсивно вызываем функцию
                for part in parts:
                    result.update(split_type(part, separators))
                break  # выходим после первого разделителя
        
        return result
````

````python
def regex_search(standarts:dict, object:str, coding_param = np.nan, return_other:bool = False):
    """Функция для поиска и замены подаваемого на вход объекта
    на стандртизованное значение"""
    # проверка типа данных входных параметров
    if not isinstance(standarts, dict):
        raise ValueError()
    
    if not isinstance(object, str):
        return coding_param
    
    # Ищем соответствие в словаре стандартных типов
    for pattern, standart_name in standarts.items():
        if re.search(pattern.lower(), object):
            return standart_name
    # параметр отвечающий за выделение неподходящих знач. в категорию other
    if return_other: 
        return 'other'
    return object
````

````python
# обрабатываем дополнительный признак этажности
word_to_num = {
        'one': 1, 'two': 2, 'three': 3, 'four': 4, 'five': 5,
        'six': 6, 'seven': 7, 'eight': 8, 'nine': 9, 'ten': 10
    }
def count_stories(story_set:set):
    """Функция для преобразования в числовой формат"""
    if isinstance(story_set, set):
        numbers = set()
        # переберем все значения множества
        for story in story_set:
            # случай когда ищем цифру 
            if re.search(r'\d+', story):
                if re.search(r'\b\d+\.\d+\b', story):# если есть float числа 
                    digits = re.findall(r'\b\d+\.\d+\b', story)
                else:
                    digits = re.findall(r'\d+', story)
                numbers.update(digits)
            # случай когда ищем число
            else: 
                # 2. Ищем словесные обозначения чисел
                for word, num in word_to_num.items():
                    # Ищем слово как целое (с границами)
                    if re.search(r'\b' + word + r'\b', story):
                        numbers.add(num)
            
            # переводим в числовой тип данных
            result = set()
            for n in numbers:
                try: 
                    result.add(float(n))
                    
                except Exception:
                    pass
                
            # проверим длину множества
            if len(result) != 0:
                return max(result)
            else:
                return np.nan
    else:
        return np.nan 

# обработаем признак этажности
PropTypeNew['new_stories'] = PropTypeNew['new_stories'].apply(count_stories)
````

#### **Обработаем признак `baths`**

`baths` - признак количества ванных комнат в объекте недвижимости. В признаке присутсвует много мусора, после определенных манипуляций разобрали и определили интерпретацию каждых из значений:
- `~` - закодированный пропуск. Не несет смысловой нагрузки.
- `0 / 0` - действительно объекты недвижимости не имеющие ванной комнаты в основном это Land (земельные участки)
- `Sq. Ft.` - аномальное значение, которое находится в количественном признаке (Удалим).
- `-- baths` - закодированный пропуск. Не несет смысловой нагрузки.
- `.../.../...` - непонятные обозначения количества ванных комнат, принято решение обозначить их пропуском.

````python
def transform_baths(number:str):
    """Функция для преобразования в правильный формат признак baths"""
    # отбросим пропуски 
    if not isinstance(number, str):
        return np.nan
    # преобразования
    trans_num = number.lower()
    trans_num = trans_num.strip()
    trans_num = trans_num.replace(',', '.')
    
    # поиск числовых значений 
    digits = re.findall(r'(\d+(?:\.\d+)?)', trans_num)
    # обработка аномальной записи .../.../... и пропусков
    if len(digits) > 2 or len(digits) == 0:
        return np.nan
    
    # преобразуем во множество
    digits_set = set(digits)
    digits_set = {float(x) for x in digits_set}
    return  max(digits_set)

house_df['baths'] = house_df.baths.apply(transform_baths)
````

#### **Обработаем признак `fireplace`**

`fireplace` - признак наличия камина. Признак имеет 1652 уникальных значения. Значения имеют различный диапазон от количественной характеристики до типа отопления и места расположения камина. Так как большая часть значений превалирует в бинарном и количественном виде, новые потенциальные признаки `тип отопления камина` и `место расположения камина` не будут созданы, учитывая кол-во данных.

Создадим новый признак `fireplace` - количество каминов. Алгоритм обработки значений следующий:
- Если строка содержит целое число  — взять его.
- Если строка содержит yes, y, fireplace (без отрицания) — присвоить 1.
- Если строка содержит no, none, not applicable, 0 — присвоить 0.

Для всех остальных текстовых описаний (включая типы каминов и их расположение) — 1 (наличие одного камина).

Пропуски в данных сразу закодируем, как отсутвие камина или кол-во каминов = 0.

````python
print(f'1) Количество уникальных значений {house_df.fireplace.nunique()}')

standart_fireplace = {
    # преобразуем прописи
    r'one': 1, r'two': 2, r'three': 3, r'four': 4, r'five': 5,
    r'six': 6, r'seven': 7, r'eight': 8, r'nine': 9, r'ten': 10,
    
    # преобразуем бинарные значения
    r'\bno\b': 0,
    r'\bnot\b': 0, 
    r'\bn/k\b': 0,
    r'\byes\b': 1,
}

def transform_fireplace(object:str):
    """Функция для преобразования признака fireplace"""
    # обработаем пропуски в явном виде (заменим на 0)
    if not isinstance(object, str):
        return 0
    
    # преобразования 
    trans_object = object.lower()
    trans_object = object.strip()
    
    # 1) преобразуем в числовой формат one, two...
    trans_object = regex_search(standart_fireplace, trans_object, coding_param=0)
    if isinstance(trans_object, int):
        return trans_object
    
    # отберем числовые значения
    digits = set(re.findall(r'\d+', trans_object))
    
    
    # если кол-во предствалено не в явном виде
    if len(digits) == 0:
        # все оставшиеся значения кодируем 1
        return 1 
    # имеются цифровые значения  
    else:
        digits = {int(x) for x in digits}
        return max(digits)
    
        
house_df['fireplace'] = house_df['fireplace'].apply(transform_fireplace)
print(f'2) Количество уникальных значений: {house_df.fireplace.nunique()}')

house_df.fireplace.describe().round(2).to_frame()
````

#### **Обработаем признак `zipcode`**

`zipcode` - почтовый индекс объекта недвижимости. 

**Как устроен ZIP‑код в США**
ZIP (Zone Improvement Plan) – это 5‑значный числовой код, где:

1. Первые три цифры обозначают сектор (область сортировки, часто соответствующую крупному городу или региону).

2. Последние две цифры уточняют зону доставки внутри сектора (почтовое отделение, район, иногда конкретную улицу).

3. Существуют также ZIP+4 (добавочные 4 цифры), которые позволяют ещё точнее идентифицировать конкретный адрес (здание, многоквартирный комплекс, почтовый ящик).

Использование полного 9‑значного кода увеличивает количество уникальных значений, что может привести к переобучению, особенно для редких комбинаций. Поэтому приведем в 5-значный формат

Укажем `zipcode` как категориальный признак в LightGBM / CatBoost / XGBoost. 

````python
print(f'1) Кол-во уникальных значений {house_df.zipcode.nunique()}')
def transform_zipcode(zipcode:str):
    "Функция для преобразования zipcode в нужный формат"
    # обработаем пропуски
    if not isinstance(zipcode, str) or ('--' in zipcode):
        return np.nan
    # базовые преобразования
    zipcode = zipcode.strip()
    
    
    # случай с 9 значным кодом 
    if '-' in zipcode:
        return int(zipcode.split('-')[0])
    
    digits = re.findall(r'\b\d+\b', zipcode)
    digits = {int(x) for x in digits}
    return max(digits)

house_df['zipcode'] = house_df.zipcode.apply(transform_zipcode)
print(f'2) Кол-во уникальных значений: {house_df.zipcode.nunique()}')
````

#### **Обработаем признак `beds`**

`beds` - признак количества спален объекта недвижимости. По своей сути признак представляет из себя количественную характеристику, однако при налаизе значений было обнаружены значения в другой единице измерения: `sqft` площадь в квадратных футах, `acres` - площадь в акрах. 

Одним из вариантом преобразования является воссоздание площади спальных комнат, взяв среднее значение площади одной спальной комнаты, можно конвертировать количественные знгачения в площадь. И наоборот из площади высчитать количество спальных комнат.

Однако данный подход не будет использован мной, так как признак площади объекта недвижимости уже пристуствует в данных, что коррелирует с площадью спальных комнат. Я сделаю более просто, не будем использовать информацию об плозади спальных комнат, закодируем их как пропуски, а на этапе заполнения пропусков заполним медианным значением исходя из среднего выборочного по типу недвижимости.

Признак содержит пропуски в неявном виде:
- `-- bd`
- `-- sqft `
- `''`

Аномальные значения: (данные об ванне)
- `baths`
- `bath`


````python
standart_beds = {
    # Пропуски в неявном виде
    r'\b--\b': np.nan,
    
    # ванная комната
    r'bath.*': np.nan,
    # площадь
    r'sqf.*': np.nan,
    r'acre.*': np.nan
}

def transform_beds(object:str):
    # обработаем пропуски 
    if not isinstance(object, str):
        return np.nan
    
    # преобразования
    trans_object = object.lower()
    trans_object = trans_object.strip()
    
    # используем рекурсивную функцию
    beds_set =  split_type(trans_object, separators=[', '])
    # проходимя по всем значениям
    
    for row in beds_set: # к примеру 1 Bath, 3 or More Bedrooms
        beds_regex = regex_search(standart_beds, row)
        if isinstance(beds_regex, str):
            digits = re.findall(r'\d+', beds_regex)
            digits = {int(x) for x in digits}
            if len(digits) != 0:
                return max(digits)
    # число не было найдено    
    return np.nan 
    
    
house_df.beds = house_df.beds.apply(transform_beds)   
display(house_df.beds.describe().round(1).to_frame())
    
    
````

#### **Обработаем признак `stories`**

`stories` - Количество этажей объекта недвижимости. В признаке встречаются значения прописи, дополнительная информация о типе недвижимости, 1.5 этажные здания. 

Дополнительную информацию не будем использовать, так как ее количество незначимо. Также при обработке признака `propertyType` был создан дополнительный информационный признак этажности. Восстановим с их помощью часть данных

````python
print(f'1) Кол-во уникальных значений: {house_df.stories.nunique()}')

standart_stories = {
    # преобразуем нецелые прописи пример: 'One and One Half'
    r'one.*and.*one.*half': 1.5,
    r'1.*1/2': 1.5,
    
    # преобразуем прописи
    r'one': 1, r'two': 2, r'three': 3, r'four': 4, r'five': 5,
    r'six': 6, r'seven': 7, r'eight': 8, r'nine': 9, r'ten': 10,
    r'tri': 3,
    
}

def transform_stories(story:str):
    """Функция для преобразования stories в нужный формат"""
    # обработаем пропуски 
    if not isinstance(story, str):
        return np.nan 
    
    # базовые преобразования
    trans_story = story.lower()
    trans_story = trans_story.strip()
    
    # разделяем на блоки ', '
    story_set = split_type(trans_story, separators=[', '])
    digits = set()
    for s in story_set:
        # преобразуем прописи
        story_regex = regex_search(standart_stories, s, coding_param=set())
        # случай когда заменили на числовой тип данных
        if isinstance(story_regex, (int, float)):
            digits.add(story_regex)
        else:
            # ищем цифры в string формате
            digits.update(re.findall(r'(\d+(?:\.\d+)?)', story_regex))
    
    # если множество не пустое тогда преобразуем в int 
    if len(digits) != 0:
        digits = {float(x) for x in digits}
        return max(digits)
    else: # наблюдения где не было числа
        return np.nan

house_df.stories = house_df.stories.apply(transform_stories)
print(f'2) Кол-во уникальных значений: {house_df.stories.nunique()}')
display(house_df.stories.describe().round(1).to_frame())
    
````

#### **Обработаем признак `price/sqft`**

`price/sqft` - это стоимость одного квадратного фута площади объекта. Признак рассчитанный по методике: `target` / `sqft` , следовательно это признак недопустимо использовать в предсказани из-за утечки дданных.

Однако можно восстановить пропуски в `sqft`:

>$$ sqft = target / price $$

Также необходимо учесть что в признаке `price/sqft` есть значения равные нулю (тип недвижимости Land к примеру).


````python
print(f'1) Кол-во уникальных значений: {house_df["price/sqft"].nunique()}')
def transform_price(price:str):
    """Функция для преобразования price/sqft в числовой тип данных"""
    # обработаем пропуски 
    if not isinstance(price, str):
        return np.nan
    
    # базовые преобразования
    trans_price = price.lower()
    trans_price = trans_price.strip()
    trans_price = trans_price.replace(',', '')
    
    digit = re.findall(r'\d+', trans_price)
    # пропуски в неявном виде
    if len(digit) == 0:
        return np.nan

    # если число найдено 
    digit = {float(x) for x in digit}
    return max(digit)
        
house_df["price/sqft"] = house_df["price/sqft"].apply(transform_price)
print(f'2) Кол-во уникальных значений: {house_df["price/sqft"].nunique()}')
display(house_df['price/sqft'].describe().round(1).to_frame())
````

````python
before_nulls = house_df.sqft.isnull().sum()
print(f'1) Кол-во пропусков в sqft: {before_nulls}')

def recover_sqft(target:float, price_sqft:float):
    # обработаем пропуски 
    if pd.isna(price_sqft):
        return np.nan 
    # обработаем нулевые значения
    if price_sqft == 0:
        return 0
    
    sqft = round(target/price_sqft)
    return sqft

# восстанавливаем только там, где sqft изначально пропущен
mask_missing = house_df['sqft'].isna()

house_df.loc[mask_missing, 'sqft'] = house_df.loc[mask_missing].apply(
    lambda row: recover_sqft(row['target'], row['price/sqft']), axis=1
)

# после этого можно удалить price/sqft, чтобы избежать утечки
house_df.drop('price/sqft', axis=1, inplace=True)

after_nulls = house_df.sqft.isnull().sum()
print(f'2) Кол-во пропусков в sqft: {after_nulls}')
print(f'3) Количество восстановленных данных: {before_nulls - after_nulls}')
````

#### **Обработаем признак `lotsize`**


- `lotsize` -  это размер земельного участка, на котором расположена недвижимость. Единицы измерений: `acres`, `sqft`. В данных присутствуют как рациональные числа к примеру *0.31 acres* так и целые числа *4,356 sqft*.

Необходимо привести в единую единицу измерений, будем переводить из `acres` в `sqft`.

$$ acres (акр) = 43560sqft (квадратным.футам)$$

````python
print(f'1) Кол-во уникальных значений: {house_df["lotsize"].nunique()}')

def transform_lotsize(sqft: str):
    """Функция для обработки lotsize в нужный числовой тип данных"""
    # обработаем пропуски
    if not isinstance(sqft, str):
        return np.nan
    
    # базовые преобразования
    trans_sqft = sqft.lower()
    trans_sqft = trans_sqft.strip()
    trans_sqft = trans_sqft.replace(',', '')
    
    # преобразуем acres в sqft
    if re.search(r'acr.*', trans_sqft):
        num = re.findall(r'(\d+(?:\.\d+)?)', trans_sqft)
        # пропуски -- acres
        if len(num) == 0: 
            return np.nan
        num = {float(x) * 43560 for x in num}
        return max(num)
    
    # обрабатываем квадратные футы
    if re.search(r'sq.*', trans_sqft):
        num = re.findall(r'\d+', trans_sqft)
        # пропуски -- sqft
        if len(num) == 0: 
            return np.nan
        num = {float(x) for x in num}
        return max(num)
    
    # обрабатываем оставшиеся значения 
    # остались либо пропуски, либо цифры без ед/измер.
    digits = re.findall(r'(\d+(?:\.\d+)?)', trans_sqft)
    # пропуски
    if len(digits) == 0:
        return np.nan 
    elif '.' in digits[0]: # значит в acres
        num = float(digits[0]) * 43560
    else:
        num = float(digits[0])
        
    return num 
        
            
        
house_df['lotsize'] = house_df.lotsize.apply(transform_lotsize)
print(f'2) Кол-во уникальных значений: {house_df["lotsize"].nunique()}')
display(house_df.lotsize.describe().round(0).to_frame())
````

#### **Восстановим данные heating и cooling**


Для восстанвления данных была написана функция `recover_func`, алгоритм действий которой прост: поиск восстановленых категорий в множестве (set), после return отобраных категорий. 

````python
def recover_func(types:set, categories:list=[], return_names:dict={}) -> set:
    """Функция предназначеная для восстовления пропусков"""
    # обработка пропусков
    if not isinstance(types, set):
        return 'unknown'
    
    # поиск восстановленных категорий
    recover_types = set()
    for type in types:
        if type in categories:
            # стандартизуем категории если указан параметр 
            if len(return_names) != 0:
                for key, value in return_names.items():
                    if key == type:
                        recover_types.add(value)
            else: # простое добавление
                recover_types.add(type)
    
    # обработка случая когда ничего не нашли 
    if len(recover_types) == 0:
        return 'unknown'
    
    return recover_types
````

````python
freq_parking = Counter()
word_to_num = {
        'one': 1, 'two': 2, 'three': 3, 'four': 4, 'five': 5,
        'six': 6, 'seven': 7, 'eight': 8, 'nine': 9, 'ten': 10
    }

def transform_parking(type: str):
    """Функция для преобразования и агрегации в категории признака parking"""
    # обработаем пропуски в явном виде
    if not isinstance(type, str): 
        freq_parking.update({'unknown'})
        return np.nan
    
    # базовые преобразования
    trans_type = type.lower()
    trans_type = trans_type.strip()
    
    # разделяем на чати 
    types_set = split_type(trans_type, [', ', ' / '])

    parking_dt = {
        'types': set(),
        'counts': set(),
    }
    
    # проходимся по каждому типу 
    for t in types_set:
        #  обработаем закодированный пропуск
        if t == '':
            freq_parking.update({'unknown'})
            return np.nan
        
        # поиск числа 
        digits = re.findall(r'\d+', t)
        if len(digits) != 0:
            digits = {int(x) for x in digits}
            parking_dt['counts'].update(digits)
            # также добавляем признак наличия парковки 
            if max(digits) != 0:
                parking_dt['types'].add('has_parking')
            else:
                parking_dt['types'].add('nonexistent')
        else:
            # стандартизируем типы 
            regex_type = regex_search(standart_parking, t, return_other=True)
            parking_dt['types'].add(regex_type)
            # если нет парковочного места добавляем 0
            if regex_type == 'nonexistent':
                parking_dt['counts'].add(0)
    
    # обработаем количество паркингов
    if len(parking_dt['counts']) == 0:
        parking_dt['counts'] = np.nan
    else:
        max_counts = max(parking_dt['counts'])
        parking_dt['counts'] = max_counts
                
    freq_parking.update(parking_dt['types'])
    return parking_dt

parking = house_df.parking.apply(transform_parking)
parking_df = pd.json_normalize(parking)
house_df['parking_types'] = parking_df.types

# в types все пропски кодируем в unknown 
parking_df.types = parking_df.types.apply(lambda x: {'unknown'} if pd.isna(x) else x)

# добавляем признак количество паркингов в датасет 
house_df['parking_counts'] = parking_df['counts']
````

#### **Функции, относящиеся к этапу очистки данных**

````python

def distribution_variable(data:pd.DataFrame, X:str, log_scale:bool = False, left:float=1.5, right:float=1.5):
    """Функция предназначенная для построения двух графиков,
    демонстрирующие распределение случайно величины. 
    Первый график - гистограмма; Второй график: - коробчатая диаграмма"""   
    desc = columns_to_plot[X][0]
    fig, axes = plt.subplots(1, 2, figsize=(15,10))
    fig.suptitle(f"Распределение: {desc}")
    # Гистограмма
    histplot = sns.histplot(
        data=data,
        x=X,
        bins=50,
        kde=True,
        palette=palette,
        ax=axes[0],
        log_scale=log_scale
    )

    # 25-ый и 75-ый квантиль
    quartile_1, quartile_3 = data[X].quantile(0.25), data[X].quantile(0.75),
    # межквартильное расстояние
    iqr = quartile_3 - quartile_1
    # верхняя и нижняя границпа ТЬюки
    lower_bound = quartile_1 - (left * iqr)
    upper_bound = quartile_3 + (right * iqr)
    
    histplot.axvline(data[X].quantile(0.5), color='k', lw=2)
    histplot.axvline(lower_bound, color='k', ls='--', lw=2)
    histplot.axvline(upper_bound, color='k', ls='--', lw=2)
    
    
    axes[0].set_title(f'Гистрограмма')
    axes[0].set_xlabel(f'{desc}')
    axes[0].set_ylabel(f'Частота')
    
    # Коробчатая диаграмма 
    sns.boxplot(
        data=data,
        x=X,
        palette=palette,
        orient='h',
        medianprops={"color": "red", "linestyle": '--'},
        ax=axes[1],
        log_scale=log_scale
    )
    axes[1].set_title(f'Коробчатая диаграмма')
    axes[1].set_xlabel(f'{desc}')
    
    plt.tight_layout()
    plt.show()
    
    
    
def outliers_iqr_mod(data, feature, left = 1.5, right = 1.5, log_scale = False): 
    # по методу Тьюки
    if (log_scale):
        x = np.log(data[feature] + 1)
    # если есть 0 значения в data[feature] то + 1 иначе ничего 
    else: x = data[feature]
    # 25-ый и 75-ый квантиль
    quartile_1, quartile_3 = x.quantile(0.25), x.quantile(0.75),
    # межквартильное расстояние
    iqr = quartile_3 - quartile_1
    # верхняя и нижняя границпа Тьюки
    lower_bound = quartile_1 - (left * iqr)
    upper_bound = quartile_3 + (right * iqr)
    
    outliers = data[(x < lower_bound) | (x > upper_bound)]
    cleaned = data[(x >= lower_bound) & (x <= upper_bound)]
    return outliers, cleaned
````

#### Признак `parking_counts`

**Для агрегированного заполнения пропусков в данном случае необходимо учитывать важный нюанс:**

Так как мы собираемся группировать данные по таким признакам как `тип парковочного места`, `тип объекта недвижимости`. Главная проблема заключается в том что один объект недвижимости может иметь несколько типов парковки или же относиться к разным типам объекта недвижимости одновременно. Данные признаки в процеесе обработки были закодированы с помощью `MultiLabelBinarizer`, соответственно, был разработан алгоритм действий по группировки таких признаков. 

Алгоритм на примере типе парковочного места (`parking_...`):
1. необходимо создать словарь содержащий медианы типов парковочных мест
2. условный блок для определения ситуаций:
    - если объект недвижимости относится только к одному типу недвижимости (кроме `parking_unknown`), то тогда берется медиана этого типа. 
    - если объект недвижимости относится к более 1 типу недвижимости (кроме `parking_unknown`), то тогда берется глобальная медиана от медиан подходящих типов недвижимости. 
    - если объект недвижимости не относится ни к 1 типу недвижимости (кроме `parking_unknown`), тогда возвращаем `NaN`. 
3. если пропуски до сих пор присутсвуют, то тогда выполняется агрегациия по типам недвижимости или заполняется медианной 


**Функции для обработки пропусков с агрегацией**

````python
def agg_func(data:pd.DataFrame, value_col:str, types_cols:list):
    """Функция для создания агрегированного словаря типов парковочного места"""
    # 1) словарь с агрегированными данными по всем типам
    grouped_dt = dict()
    for type in types_cols:
        # соответсвует определенному типу парковочного места
        mask = data[type] == 1 
        value = data[mask].groupby(type)[value_col].median().iloc[0]
        grouped_dt[type] = value
        
    return grouped_dt
````

Функция по восстановлению агрегированных значений по типу недвижимости или паркингу 

````python
def recover_func(types_set, preff:str, agg_dt:dict) -> float:
    "Функция условный блок по агрегированному восстановлению пропусков"
    
    # обработаем 3 случай: когда тип парковочного места не указан. 
    if pd.isna(types_set):
        return np.nan 

    # обработаем 1 случай: когда только 1 тип парковочного места
    if len(types_set) == 1:
        # обрабатываем исключение 
        if 'unknown' in types_set:
            return np.nan 
        else: 
            for type in types_set:
                type_median = agg_dt[preff + type]
                # случаи когда агрегация несработала и медианого значения в целом нет
                if pd.isna(type_median):
                    return np.nan 
                else:
                    return type_median

    # обработаем 2 случай: когда более 1 типа парковочного места
    if len(types_set) > 1:
        medians_lst = []
        for type in types_set:
            # обработаем исключение 
            if 'unknown' in type:
                continue
            
            type_median = agg_dt[preff + type]
            # обработаем случай когда медианного значения в целом нет
            if pd.isna(type_median):
                continue
            else:
                medians_lst.append(type_median)
        
        return np.median(medians_lst)
````

**Восстановим данные основываясь на типе парковочного места**

````python
# признаки относящиеся к типам парковочного места 
mask = house_df.columns.str.startswith('parking')
parking_cols = house_df.columns[mask].drop(['parking_counts',
                                            'parking_unknown',
                                            'parking_types']) 

# 1) словарь с агрегированными значениями
parking_dt = agg_func(house_df, 'parking_counts', parking_cols)

mask_nulls = house_df.parking_counts.isna()  
before_nulls = mask_nulls.sum() 
# 2) восстанавливаем данные (условный блок)
house_df.loc[mask_nulls, 'parking_counts'] = house_df.loc[
    mask_nulls,'parking_types'].apply(
                    lambda s: recover_func(s, 'parking_', parking_dt))
    
after_nulls = house_df.parking_counts.isnull().sum()
print(f"Кол-во восстановленных данных: {before_nulls - after_nulls}")
````

**Восстановим данные основываясь на типе объекта недвижимости (пункт 3)**

````python
# признаки относящиеся к типам объекта недвижимости
mask = house_df.columns.str.startswith('type')
houses_cols = house_df.columns[mask]

# 1) создаем словарь с агрегацией типов недвижимости 
houses_types = agg_func(house_df, 'parking_counts', houses_cols)

mask_nulls = house_df.parking_counts.isna()  
before_nulls = mask_nulls.sum() 
# 2) восстанавливаем данные
house_df.loc[mask_nulls,'parking_counts'] = house_df.loc[
    mask_nulls,'PropTypeNew'].apply(
                    lambda s: recover_func(s, 'type_', houses_types))
    
after_nulls = house_df.parking_counts.isnull().sum()
print(f"Кол-во восстановленных данных: {before_nulls - after_nulls}")
````

#### Признаки связанные с образованием (`average_rating_schools`, `max_rating_schools`, `has_high_school`, `has_middle_school`, `has_junior_school`, `min_distance_school`, `average_distance_school`)

Пропуски заполняем нулевым значением 



````python
# 1) average_rating_school
nulls = house_df['average_rating_schools'].isnull()
# индикатор наличия школ
has_school = house_df[[
    'has_junior_school',
    'has_middle_school',
    'has_high_school']].sum(axis=1) > 0 

# медианна по всей выборке 
global_median = house_df.average_rating_schools.median()
house_df.loc[nulls & has_school.index,'average_rating_schools'] = house_df.loc[
                                nulls & has_school.index,
                                'average_rating_schools'].apply(lambda x: global_median)

# оставшиеся значения заполняем нулем 
house_df['average_rating_schools'].fillna(0, inplace=True)

# ---------------------------------------------------------------

# 2) max_rating_schools
nulls = house_df['max_rating_schools'].isnull()
# индикатор наличия школ
has_school = house_df[[
    'has_junior_school',
    'has_middle_school',
    'has_high_school']].sum(axis=1) > 0 

# медианна по всей выборке 
global_median = house_df.max_rating_schools.median()
house_df.loc[nulls & has_school.index,'max_rating_schools'] = house_df.loc[
                                nulls & has_school.index,
                                'max_rating_schools'].apply(lambda x: global_median)

# оставшиеся значения заполняем нулем 
house_df['max_rating_schools'].fillna(0, inplace=True)

# ---------------------------------------------------------------

# 3) признаки наличия школы определенного уровня образования
features = ['has_junior_school',
            'has_middle_school',
            'has_high_school']
# заполняем False
for col in features:
    nulls = house_df[col].isna()
    house_df.loc[nulls, col] = False

# приводим к типу данных bool
house_df[features] = house_df[features].astype(bool)  

# ---------------------------------------------------------------

# 4) признаки дистанции до школ
features = ['min_distance_school',
            'average_distance_school']
# заполняем медианным значением
for col in features:
    global_median = house_df[col].median()
    nulls = house_df[col].isna()
    house_df.loc[nulls, col] = global_median
````
