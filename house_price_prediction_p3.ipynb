#### **Выбор метрики для решения задачи прогнозирования цены объекта недвижимости**

Обоснование и выбор метрик для оценки качества модели прогнозирования цены недвижимости

В рамках проекта по предсказанию стоимости объектов недвижимости (регрессионная задача) ключевым этапом является выбор метрики, позволяющей объективно оценить качество модели. Целевая переменная – цена объекта – имеет правостороннюю асимметрию (наличие дорогих объектов, не являющихся аномалиями) и измеряется в абсолютных денежных единицах (рублях). В связи с этим предъявляются следующие требования к метрикам:

1. Интерпретируемость для бизнес-заказчика (ошибка должна быть выражена в тех же единицах, что и цена).
2. Устойчивость к выбросам (дорогие объекты не должны искажать общую оценку качества).
3. Возможность сравнения моделей с разными наборами признаков.

На основе анализа достоинств и недостатков различных метрик были выбраны следующие показатели.

**Основная метрика** – `средняя абсолютная ошибка (MAE, Mean Absolute Error)`. Данная метрика вычисляется как среднее арифметическое модулей разностей между предсказанными и реальными значениями цены. MAE обладает следующими преимуществами:
- Измеряется в тех же единицах, что и целевая переменная, что позволяет легко интерпретировать результат («в среднем модель ошибается на X долларов»).
- Не чувствительна к выбросам, поскольку не возводит ошибку в квадрат. Это особенно важно для нашего датасета, где присутствуют объекты с высокой стоимостью.
- Является стандартной метрикой в задачах регрессии с асимметричным распределением целевой переменной.

**Дополнительная метрика** – `средняя абсолютная процентная ошибка (MAPE, Mean Absolute Percentage Error)`. Она позволяет оценить относительную точность предсказаний в процентах. Преимущества MAPE:
- Легкая интерпретация («модель ошибается в среднем на Y%»).
- Возможность сравнения качества модели на разных датасетах или с разными масштабами цен.
В нашем исследовании MAPE используется как вспомогательная метрика для оценки относительной погрешности, однако следует учитывать, что она может быть нестабильна при очень малых истинных значениях цены.

**Метрика для сравнения моделей** – `коэффициент детерминации R² (R-squared)`. Она показывает, какую долю дисперсии целевой переменной объясняет модель. Значение R² лежит в диапазоне (-∞, 1], причём значение 1 означает идеальное предсказание, а 0 – модель не лучше, чем предсказание среднего. R² удобна для сравнения различных наборов признаков и алгоритмов, поскольку она не зависит от масштаба данных. В рамках проекта удовлетворительным считается значение R² выше 0.7.

Отказ от использования MSE и RMSE обосновывается их высокой чувствительностью к выбросам (квадратичный штраф за большие ошибки), что при наличии дорогих объектов недвижимости может привести к неадекватной оценке качества модели. Кроме того, квадратичные метрики менее интерпретируемы для бизнеса, так как единицы измерения (рубли в квадрате) не имеют практического смысла.

Таким образом, в работе принят следующий набор метрик:
- **MAE ($) – основная метрика точности.**
- **MAPE (%) – дополнительная относительная метрика.**
- **R² – метрика для сравнения объясняющей способности моделей.**

#### **Функции для визуализации результатов**

````python
def shap_summary_plot(model, X: pd.DataFrame, **kwargs):
    """
    Глобальная важность признаков с указанием знака влияния (SHAP summary plot).

    Args:
        model: обученная модель на основе деревьев (RandomForest, XGBoost, LightGBM, CatBoost).
        X (pd.DataFrame): данные для объяснения.
        **kwargs: дополнительные параметры для shap.summary_plot (например, show=False).
    """
    explainer = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(X)
    shap.summary_plot(shap_values, X, **kwargs, )
    plt.tight_layout()
    plt.show()


def shap_waterfall_plot(model, X: pd.DataFrame, index: int, **kwargs):
    """
    Waterfall plot для одного наблюдения (водопад вкладов признаков).

    Args:
        model: обученная модель на основе деревьев.
        X (pd.DataFrame): данные.
        index (int): индекс строки.
        **kwargs: параметры shap.waterfall_plot (например, max_display).
    """
    explainer = shap.TreeExplainer(model)
     # Берём DataFrame с одной строкой по метке
    X_row = X.loc[[index]]
    shap_values = explainer.shap_values(X_row)

    # Формируем Explanation объект для waterfall
    exp = shap.Explanation(values=shap_values[0],
                           base_values=explainer.expected_value,
                           data=X.loc[index],
                           feature_names=X.columns.tolist())
    shap.waterfall_plot(exp, **kwargs)
    plt.tight_layout()
    plt.show()
    
    
def plot_actual_vs_predicted(y_true: np.ndarray, y_pred: np.ndarray) -> None:
    """
    Диаграмма рассеяния фактических vs предсказанных значений с линией идеала.

    Args:
        y_true: истинные значения
        y_pred: предсказанные значения
        title: заголовок графика (опционально)
    """
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=y_true, y=y_pred, alpha=0.6, edgecolor='none')
    min_val = min(y_true.min(), y_pred.min())
    max_val = max(y_true.max(), y_pred.max())
    plt.plot([min_val, max_val], [min_val, max_val], 'r--', lw=2, label='Идеальная линия')
    plt.xlabel('Фактические значения')
    plt.ylabel('Предсказанные значения')
    plt.title('Фактические vs предсказанные значения')
    plt.legend()
    plt.tight_layout()
    plt.show()
    
    
def plot_learning_curves(model, metric_name: str = 'RMSE') -> None:
    """
    Кривые обучения для градиентного бустинга (по итерациям).
    Модель должна быть обучена с параметром `eval_set` и иметь атрибут `evals_result_`.

    Args:
        model: обученная модель XGBoost/LightGBM/CatBoost с evals_result_
        metric_name: название метрики для подписи оси
    """
    evals_result = model.evals_result_
    train_key = list(evals_result.keys())[0]
    # берём второй ключ, если есть
    keys = list(evals_result.keys())
    val_key = keys[1] if len(keys) > 1 else train_key

    # Извлекаем историю метрик
    train_metrics = list(evals_result[train_key].values())[0]
    val_metrics = list(evals_result[val_key].values())[0]
    iterations = range(1, len(train_metrics) + 1)

    plt.figure(figsize=(10, 6))
    plt.plot(iterations, train_metrics, label=f'Обучающая выборка ({metric_name})')
    plt.plot(iterations, val_metrics, label=f'Валидационная выборка ({metric_name})')
    plt.xlabel('Число итераций (деревьев)')
    plt.ylabel(metric_name)
    plt.title(f'Кривые обучения: {metric_name}')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    
````

Перед подбором гипперпараметров необходимо закодировать текстовые категориальные признаки `state`, `city` с помощью `OneHotEncoding`, после стандартизируем. Подберем гипперпараметры с помощью `Optuna`. Обучим модель случайного леса. 

````python
# копируем исходный датасет 
df_encoded = house_df.copy()

# частотная выборка
state_top_15 = df_encoded.state.value_counts().head(15).index
city_top_15 = df_encoded.city.value_counts().head(15).index
# заменяем на others 
df_encoded.state = df_encoded.state.apply(lambda x: x if x in state_top_15 else 'other')
df_encoded.city = df_encoded.city.apply(lambda x: x if x in city_top_15 else 'other')

# кодируем onehot state
state_cols = pd.get_dummies(df_encoded.state, prefix='state_').astype('category')
# кодируем onehot city
city_cols = pd.get_dummies(df_encoded.city, prefix='city_').astype('category')
# объединяем 
df_encoded = pd.concat([df_encoded, state_cols, city_cols], axis=1)
df_encoded.drop(['zipcode', 'state', 'city', 'target_log'], axis=1, inplace=True)

# создаем новые выборки для RandomForest и XGBOOST
X_train_enc = df_encoded.loc[train_index]
X_test_enc = df_encoded.loc[test_index]
X_valid_enc = df_encoded.loc[valid_index]

stand_cols = X_train_enc.columns
# стандартизируем данные
scaler = StandardScaler()
# преобразуем
X_train_enc[stand_cols] = scaler.fit_transform(X_train_enc[stand_cols])
X_valid_enc[stand_cols] = scaler.transform(X_valid_enc[stand_cols])
X_test_enc[stand_cols] = scaler.transform(X_test_enc[stand_cols])
````

#### **Функция для подбора гипперпараметров модели `RandomForestRegressor`**

````python
def rf_params(trial):
    """Целевая функция для Optuna: обучает RandomForest с предложенными параметрами,
    возвращает метрику на валидации (минимизируем MAE)."""
    
    # Пространство поиска гиперпараметров
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 300, 1000, step=100),
        'max_depth': trial.suggest_int('max_depth', 5, 20),
        'min_samples_split': trial.suggest_int('min_samples_split', 2, 40),
        'min_samples_leaf': trial.suggest_int('min_samples_leaf', 1, 30),
        'max_features': 0.3,
        'ccp_alpha': trial.suggest_float('ccp_alpha', 0.0, 0.05),
        'bootstrap': True,
        'n_jobs': -1,
        'random_state': 42,
    }
    
    # Создаём модель
    model = RandomForestRegressor(**params)
    
    # Обучаем с валидационной выборкой
    model.fit(X_train_enc, y_train)
    
    # Предсказание на валидации
    preds = model.predict(X_valid_enc)
    # Метрика для оптимизации :
    mae = mean_absolute_error(np.exp(y_valid), np.exp(preds))
    return mae 
````

````python
# проверка наличия подобранных параметров
if (not os.path.exists('data/best_params_rf.pkl')) or (not os.path.exists('data/study_rf.pkl')):
    # Создаём исследование
    study = optuna.create_study(direction='minimize', study_name='randomforest_tuning')
    # Запускаем оптимизацию 
    study.optimize(rf_params, n_trials=200, show_progress_bar=True)

    # Вывод лучших параметров и результата
    print("Лучшие параметры:", study.best_params)
    print("Лучшее значение MAE на валидации:", study.best_value)
    best_params = study.best_params

    # сериализируем лучшие параметры 
    with open('data/optuna/best_params_rf.pkl', 'wb') as output:
        pickle.dump(best_params, output)
    # сериализируем объект study
    with open('data/optuna/study_rf.pkl', 'wb') as output:
        pickle.dump(study, output)
else:
    # десериализируем лучшие параметры 
    with open('data/optuna/best_params_rf.pkl', 'rb') as file:
        best_params = pickle.load(file)
    # десериализируем объект study
    with open('data/optuna/study_rf.pkl', 'rb') as file:
        study = pickle.load(file)
        
    # Вывод лучших параметров и результата
    print("Лучшие параметры:", study.best_params)
    print("Лучшее значение MAE на валидации:", study.best_value)
````

#### **Функция для подбора гипперпараметров модели `CatBoostRegressor`**

````python
def cat_params(trial):
    """Целевая функция для Optuna: обучает CatBoost с предложенными параметрами,
    возвращает метрику на валидации (минимизируем MAE)."""
    
    # Пространство поиска гиперпараметров
    params = {
        'iterations': 300,
        'learning_rate': trial.suggest_float('learning_rate', 0.05, 0.15, log=True),
        'depth': trial.suggest_int('depth', 12, 15),
        'l2_leaf_reg': trial.suggest_float('l2_leaf_reg', 2, 6, log=True),
        'min_data_in_leaf': trial.suggest_int('min_data_in_leaf', 50, 75),
        'random_strength': 1,
        'bagging_temperature': 1,
        'subsample': trial.suggest_float('subsample', 0.7, 0.9),
        'colsample_bylevel': 0.85,
        # Регуляризация модели
        'model_size_reg': 0.2,
        # Ранняя остановка и контроль переобучения
        'early_stopping_rounds': 50,
        'use_best_model': True,        
        'random_seed': 42
    }
    
    # Создаём модель
    model = CatBoostRegressor(**params)
    
    # Обучаем с валидационной выборкой
    model.fit(
        X_train, y_train,
        cat_features=['state', 'zipcode', 'city'],
        eval_set=(X_valid, y_valid),
        verbose=False
    )
    
    # Предсказание на валидации
    preds = model.predict(X_valid)
    # Метрика для оптимизации MAE:
    mae = mean_absolute_error(np.exp(y_valid), np.exp(preds))
    return mae 
````

````python
if (not os.path.exists('data/best_params_catboost.pkl')) or (not os.path.exists('data/study_catboost.pkl')): 
    # Создаём исследование
    study = optuna.create_study(direction='minimize', study_name='catboost_tuning', pruner=MedianPruner(n_startup_trials=5, n_warmup_steps=30))

    # Запускаем оптимизацию 
    study.optimize(cat_params, n_trials=100, show_progress_bar=True)

    # Вывод лучших параметров и результата
    print("Лучшие параметры:", study.best_params)
    print("Лучшее значение MAE на валидации:", study.best_value)
    best_params = study.best_params
    
    # сериализируем лучшие параметры 
    with open('data/optuna/best_params_catboost.pkl', 'wb') as output:
        pickle.dump(best_params, output)
    # сериализируем объект study 
    with open('data/optuna/study_catboost.pkl', 'wb') as output:
        pickle.dump(study, output)
else:
    # десериализируем лучшие параметры 
    with open('data/optuna/best_params_catboost.pkl', 'rb') as file:
        best_params = pickle.load(file)
    # десериализируем объект study 
    with open('data/optuna/study_catboost.pkl', 'rb') as file:
        study = pickle.load(file)
        
    # Вывод лучших параметров и результата
    print("Лучшие параметры:", study.best_params)
    print("Лучшее значение MAE на валидации:", study.best_value)
    best_params = study.best_params
````

#### **Функция для подбора гипперпараметров `XGBOOST`**

````python
def xgb_params(trial):
    """Целевая функция для Optuna: обучает XGBoost с предложенными параметрами,
    возвращает MAE на валидации (в исходном масштабе, через np.exp)."""
    
    params = {
        # Количество деревьев 
        'n_estimators': trial.suggest_int('n_estimators', 500, 4000, step=100),
        # Скорость обучения 
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        # Максимальная глубина дерева 
        'max_depth': trial.suggest_int('max_depth', 4, 16),
        # L2 регуляризация на веса 
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-2, 10.0, log=True),
        # L1 регуляризация 
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-2, 10.0, log=True),
        # Минимальный суммарный вес в листе 
        'min_child_weight': trial.suggest_int('min_child_weight', 5, 100),
        # Доля признаков, используемая для каждого дерева
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.5, 1.0),
        # Доля признаков, используемая для каждого уровня дерева
        'colsample_bylevel': trial.suggest_float('colsample_bylevel', 0.5, 1.0),
        # Доля объектов для каждого дерева (аналог subsample)
        'subsample': trial.suggest_float('subsample', 0.5, 1.0),
        # Минимальное улучшение loss для дальнейшего разделения узла 
        'gamma': trial.suggest_float('gamma', 0.0, 5.0),
        'early_stopping_rounds': 50,
        'random_state': 42,
        'verbosity': 0,
        'n_jobs': -1
    }
    
    model = xgb.XGBRegressor(**params)
    # Обучение с валидационной выборкой и ранней остановкой
    model.fit(
        X_train_enc, y_train,
        eval_set=[(X_valid_enc, y_valid)],
        verbose=False
    )
    
    # Предсказание на валидации (лучшая модель по валидационной метрике)
    preds = model.predict(X_valid_enc)
    # Если вы работаете с логарифмированной целевой переменной
    mae = mean_absolute_error(np.exp(y_valid), np.exp(preds))    
    return mae

````

#### **Функция для подбора гипперпараметров LIGHTGBM**

````python
def lgb_params(trial):
    """Целевая функция для Optuna: обучает LightGBM с предложенными параметрами,
    возвращает MAE на валидации (в исходном масштабе через np.exp)."""
    
    params = {
        # Количество деревьев 
        'n_estimators': trial.suggest_int('n_estimators', 1000, 4000, step=100),
        # Скорость обучения 
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        # Максимальная глубина дерева 
        'max_depth': trial.suggest_int('max_depth', 3, 16),
        # L2 регуляризация на веса 
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-2, 10.0, log=True),
        # L1 регуляризация
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-2, 10.0, log=True),
        # Минимальное количество данных в листе 
        'min_child_samples': trial.suggest_int('min_child_samples', 5, 100),
        # Доля признаков, используемая для каждого дерева 
        'feature_fraction': trial.suggest_float('feature_fraction', 0.5, 1.0),
        # Доля объектов для каждого дерева 
        'bagging_fraction': trial.suggest_float('bagging_fraction', 0.5, 1.0),
        # Частота bagging 
        'bagging_freq': trial.suggest_int('bagging_freq', 1, 10),
        # Минимальное улучшение loss для разделения
        'min_split_gain': trial.suggest_float('min_split_gain', 0.0, 5.0),
        # Количество листьев
        'num_leaves': trial.suggest_int('num_leaves', 10, 255),
        # Ранняя остановка 
        'early_stopping_rounds': 50,
        'random_state': 42,
        'verbose': -1,          
        'n_jobs': -1
    }
    
    # Создаём модель
    model = lgb.LGBMRegressor(**params)
    
    # Обучение с валидационной выборкой и ранней остановкой
    model.fit(
        X_train, y_train,
        eval_set=[(X_valid, y_valid)],
        categorical_feature=['state', 'zipcode', 'city'],  
    )
    
    # Предсказание на валидации
    preds = model.predict(X_valid)
    # Если вы работаете с логарифмированной целевой переменной
    mae = mean_absolute_error(np.exp(y_valid), np.exp(preds))
    return mae
````

````python
if (not os.path.exists('data/best_params_lgb.pkl')) or (not os.path.exists('data/study_lgb.pkl')): 
    # Создаём исследование
    study = optuna.create_study(direction='minimize', study_name='lgb_tuning')

    # Запускаем оптимизацию 
    study.optimize(lgb_params, n_trials=500, show_progress_bar=True)

    # Вывод лучших параметров и результата
    print("Лучшие параметры:", study.best_params)
    print("Лучшее значение MAE на валидации:", study.best_value)
    best_params = study.best_params
    
    # сериализируем лучшие параметры 
    with open('data/optuna/best_params_lgb.pkl', 'wb') as output:
        pickle.dump(best_params, output)
    # сериализируем объект study 
    with open('data/optuna/study_lgb.pkl', 'wb') as output:
        pickle.dump(study, output)
else:
    # десериализируем лучшие параметры 
    with open('data/optuna/best_params_lgb.pkl', 'rb') as file:
        best_params = pickle.load(file)
    # десериализируем объект study 
    with open('data/optuna/study_lgb.pkl', 'rb') as file:
        study = pickle.load(file)
````

````python
best_params['n_estimators'] = 4000  # можно увеличить iterations для финальной модели
best_params['verbose'] = -1
best_params['early_stopping_rounds'] = 50

lgb_model = lgb.LGBMRegressor(**best_params, random_seed=42, n_jobs=-1)
lgb_model.fit(
        X_train, y_train,
        eval_set=[(X_valid, y_valid)],
        categorical_feature=['state', 'zipcode', 'city'],  
        callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)]  # альтернативный способ
    )

# сериализуем модель 
with open('models/lgb.pkl', 'wb') as output:
    pickle.dump(lgb_model, output)
````
